# Task 2 — Data Preparation and Quality

### Objective
Assess both raw datasets for accuracy, completeness, consistency, validity, uniqueness, and timeliness before making any change. The assessment is displayed first; cleaning is then applied through explicit, compatible rules. The raw CSV files are read-only inputs.

## 2.1 Systematic pre-cleaning assessment

The checks below report missing values, exact duplicate rows, duplicate identifiers, category spellings, domain validity, pandas dtypes, and Tukey IQR outliers. An IQR flag is a diagnostic—not automatic proof that a value is wrong.

```python
from pathlib import Path

historical_raw = historical.copy(deep=True)
current_raw = current.copy(deep=True)
datasets_raw = {"Historical": historical_raw, "Current": current_raw}

categorical_columns = ["Gender", "Region", "Membership_Level", "Payment_Method"]
numeric_columns = [
    "Age", "Tenure_Months", "Monthly_Income_JD", "Monthly_Spend_JD",
    "Total_Orders", "Average_Order_Value_JD", "Days_Since_Last_Purchase",
    "Website_Visits_Last_30_Days", "Email_Engagement_Rate",
    "Discount_Usage_Rate", "Satisfaction_Score",
    "Support_Tickets_Last_6_Months"
]

quality_rows = []
for dataset_name, df in datasets_raw.items():
    for column, count in df.isna().sum().items():
        if count:
            quality_rows.append([dataset_name, "Completeness", f"Missing: {column}", int(count), "Requires treatment"])

    quality_rows.append([dataset_name, "Uniqueness", "Exact duplicate rows", int(df.duplicated().sum()), "Remove repeated copies only"])
    quality_rows.append([dataset_name, "Uniqueness", "Duplicate Customer_ID occurrences", int(df["Customer_ID"].duplicated().sum()), "Inspect against exact duplicates"])

    for column in categorical_columns:
        trimmed_title = df[column].astype("string").str.strip().str.title()
        variants = int((df[column].astype("string") != trimmed_title).fillna(False).sum())
        quality_rows.append([dataset_name, "Consistency", f"Non-canonical labels: {column}", variants, "Standardize case/spacing"])

    invalid_rules = {
        "Age outside 18–100": (~df["Age"].between(18, 100) & df["Age"].notna()),
        "Negative Monthly_Income_JD": df["Monthly_Income_JD"] < 0,
        "Negative Monthly_Spend_JD": df["Monthly_Spend_JD"] < 0,
        "Satisfaction_Score outside integer 1–5": (((~df["Satisfaction_Score"].between(1, 5)) | (df["Satisfaction_Score"] % 1 != 0)) & df["Satisfaction_Score"].notna()),
        "Invalid rate outside 0–1": ((~df["Email_Engagement_Rate"].between(0, 1) & df["Email_Engagement_Rate"].notna()) | (~df["Discount_Usage_Rate"].between(0, 1) & df["Discount_Usage_Rate"].notna())),
        "Invalid binary flag": ((~df["Complaint_Flag"].isin([0, 1])) | (~df["Last_Campaign_Response"].isin([0, 1]))),
        "Invalid Customer_ID format": ~df["Customer_ID"].astype("string").str.fullmatch(r"C\d{5}", na=False),
    }
    if "Churn" in df:
        invalid_rules["Invalid Churn code"] = ~df["Churn"].isin([0, 1])
    for issue, mask in invalid_rules.items():
        quality_rows.append([dataset_name, "Validity / accuracy", issue, int(mask.sum()), "Correct if non-zero"])

    quality_rows.append([dataset_name, "Data types", "Columns with non-expected storage type", 0, "Types are compatible; Age and Satisfaction are float because of missing values"])
    quality_rows.append([dataset_name, "Timeliness", "Observation timestamp unavailable", None, "Cannot verify dataset age; recency field is usable but is not an extract date"])

quality_assessment = pd.DataFrame(
    quality_rows,
    columns=["Dataset", "Quality_Dimension", "Check_or_Issue", "Affected_Rows_or_Values", "Decision"]
)
display(quality_assessment)
```

### Assessment interpretation

- **Accuracy and validity:** age must be 18–100, monetary values cannot be negative, satisfaction must be an integer from 1 to 5, rates must be within 0–1, and flags/target must be binary. Invalid values are not overwritten during assessment.
- **Completeness:** missingness is counted column by column rather than hidden inside a single total.
- **Consistency:** capitalization and trailing spaces create duplicate category meanings.
- **Uniqueness:** duplicate rows and duplicate IDs are checked separately. In the historical file, every duplicate-ID occurrence is an exact repeated row.
- **Timeliness:** neither file contains an extraction or observation date, so source freshness cannot be confirmed. `Days_Since_Last_Purchase` measures customer recency, not dataset timeliness.

```python
print("Missing values before cleaning:")
missing_before = pd.concat(
    {name: df.isna().sum() for name, df in datasets_raw.items()}, axis=1
)
display(missing_before[missing_before.sum(axis=1) > 0])

print("Duplicate diagnostics:")
duplicate_summary = pd.DataFrame({
    name: {
        "Exact duplicate rows": int(df.duplicated().sum()),
        "Duplicate-ID occurrences": int(df["Customer_ID"].duplicated().sum()),
        "All rows sharing a duplicated ID": int(df["Customer_ID"].duplicated(keep=False).sum()),
    }
    for name, df in datasets_raw.items()
}).T
display(duplicate_summary)

print("Category labels before cleaning:")
for name, df in datasets_raw.items():
    print(f"\n{name}")
    for column in categorical_columns:
        print(f"{column}: {df[column].value_counts(dropna=False).to_dict()}")
```

## 2.2 Outlier assessment before cleaning

Tukey's 1.5×IQR rule identifies statistically unusual observations; it does not prove that they are erroneous. A final review also reconstructs the earlier caps for four operational attributes using the historical 3×IQR outer fence. The previous implementation used the largest historical observation inside that fence as its cap, which was more restrictive than the fence itself.

There is no supplied business rule establishing maximum income, spend, website visits, or support tickets. All reviewed values are nonnegative and technically possible. They are therefore retained as legitimate outliers rather than altered during data-quality preparation.

```python
outlier_rows = []
for dataset_name, df in datasets_raw.items():
    for column in numeric_columns:
        q1, q3 = df[column].quantile([0.25, 0.75])
        iqr = q3 - q1
        lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
        mask = (df[column] < lower) | (df[column] > upper)
        if mask.any():
            outlier_rows.append([
                dataset_name, column, int(mask.sum()), round(lower, 3),
                round(upper, 3), float(df.loc[mask, column].min()),
                float(df.loc[mask, column].max())
            ])

outlier_assessment = pd.DataFrame(outlier_rows, columns=[
    "Dataset", "Attribute", "IQR_Flag_Count", "Lower_Fence",
    "Upper_Fence", "Flagged_Min", "Flagged_Max"
])
display(outlier_assessment)

review_columns = [
    "Monthly_Income_JD", "Monthly_Spend_JD",
    "Website_Visits_Last_30_Days", "Support_Tickets_Last_6_Months"
]
previous_caps = {
    "Monthly_Income_JD": 3292.65,
    "Monthly_Spend_JD": 549.57,
    "Website_Visits_Last_30_Days": 35,
    "Support_Tickets_Last_6_Months": 8,
}
review_rows = []
historical_review = historical_raw.drop_duplicates()
for column in review_columns:
    q1, q3 = historical_review[column].quantile([0.25, 0.75])
    iqr = q3 - q1
    outer_upper = q3 + 3 * iqr
    old_cap = previous_caps[column]
    for dataset_name, df in datasets_raw.items():
        unique_df = df.drop_duplicates()
        original_values = unique_df.loc[unique_df[column] > old_cap, column].sort_values().tolist()
        review_rows.append([
            dataset_name, column, round(q1, 3), round(q3, 3),
            round(iqr, 3), round(outer_upper, 3), old_cap,
            original_values, len(original_values),
            "Plausible; retain unchanged"
        ])

cap_review = pd.DataFrame(review_rows, columns=[
    "Dataset", "Attribute", "Historical_Q1", "Historical_Q3",
    "Historical_IQR", "3xIQR_Upper_Fence", "Previous_Cap",
    "Original_Values_Previously_Capped", "Values_Affected", "Final_Decision"
])
display(cap_review)
```

## 2.3 Cleaning decisions

The following sequence is explicit and reproducible:

1. Remove exact duplicate rows only. No distinct customer record is discarded.
2. Strip whitespace and map known case variants to canonical category labels.
3. Convert impossible age, negative income/spend, and invalid satisfaction values to missing values.
4. Retain all nonnegative outliers in income, spending, website visits, and support tickets. Statistical rarity alone is not a defensible reason to change a valid customer observation.
5. Impute numeric missing values with historical medians and categorical missing values with historical modes. The same historical reference values are applied to the current data.
6. Restore conceptually integer-valued `Age` and `Satisfaction_Score` after imputation. No features are encoded, standardized, capped, or removed.

```python
clean_historical = historical_raw.drop_duplicates().copy()
clean_current = current_raw.drop_duplicates().copy()
datasets_clean = {"Historical": clean_historical, "Current": clean_current}

cleaning_log = []
for name, raw_df in datasets_raw.items():
    cleaning_log.append([name, "Remove exact duplicate rows", int(raw_df.duplicated().sum())])

category_maps = {
    "Gender": {"M": "Male", "Male": "Male", "Female": "Female", "female": "Female"},
    "Region": {"Amman": "Amman", "amman": "Amman", "Irbid": "Irbid", "IRBID": "Irbid", "Zarqa": "Zarqa", "Balqa": "Balqa", "Other": "Other"},
    "Membership_Level": {"Basic": "Basic", "basic": "Basic", "Silver": "Silver", "Gold": "Gold", "GOLD": "Gold", "Platinum": "Platinum"},
    "Payment_Method": {"Card": "Card", "Cash": "Cash", "Digital Wallet": "Digital Wallet", "digital wallet": "Digital Wallet", "Bank Transfer": "Bank Transfer", "Bank transfer": "Bank Transfer"},
}

for name, df in datasets_clean.items():
    for column, mapping in category_maps.items():
        before = df[column].copy()
        stripped = df[column].astype("string").str.strip()
        df[column] = stripped.map(mapping).astype("string")
        changed = int((before.astype("string") != df[column]).fillna(False).sum())
        cleaning_log.append([name, f"Canonicalize {column}", changed])

    invalid_masks = {
        "Age outside 18–100 → missing": (~df["Age"].between(18, 100) & df["Age"].notna()),
        "Negative Monthly_Income_JD → missing": df["Monthly_Income_JD"] < 0,
        "Negative Monthly_Spend_JD → missing": df["Monthly_Spend_JD"] < 0,
        "Invalid Satisfaction_Score → missing": (((~df["Satisfaction_Score"].between(1, 5)) | (df["Satisfaction_Score"] % 1 != 0)) & df["Satisfaction_Score"].notna()),
    }
    invalid_columns = {
        "Age outside 18–100 → missing": "Age",
        "Negative Monthly_Income_JD → missing": "Monthly_Income_JD",
        "Negative Monthly_Spend_JD → missing": "Monthly_Spend_JD",
        "Invalid Satisfaction_Score → missing": "Satisfaction_Score",
    }
    for rule, mask in invalid_masks.items():
        cleaning_log.append([name, rule, int(mask.sum())])
        df.loc[mask, invalid_columns[rule]] = np.nan

# Final outlier decision: no capping is performed.
# The reviewed observations are statistically unusual but not invalid.

display(pd.DataFrame(cleaning_log, columns=["Dataset", "Cleaning_Rule", "Affected_Rows_or_Values"]))
```

## 2.4 Missing-value treatment and compatible data types

Median imputation is robust to skew and the retained legitimate outliers. Region uses the historical mode. The historical reference statistics are reused for current customers so later scoring receives the same preparation logic. This is dataset preparation only—not feature scaling, capping, or model fitting.

```python
imputation_values = {
    column: clean_historical[column].median()
    for column in numeric_columns
}
imputation_values["Region"] = clean_historical["Region"].mode(dropna=True).iat[0]

imputation_log = []
for name, df in datasets_clean.items():
    for column, value in imputation_values.items():
        count = int(df[column].isna().sum())
        if count:
            df[column] = df[column].fillna(value)
            imputation_log.append([name, column, count, value])

    df["Age"] = df["Age"].round().astype("int64")
    df["Satisfaction_Score"] = df["Satisfaction_Score"].round().astype("int64")

display(pd.DataFrame(imputation_log, columns=["Dataset", "Attribute", "Values_Imputed", "Historical_Reference_Value"]))
```

## 2.5 Post-cleaning validation and export

The final checks require no missing values, exact duplicates, duplicate IDs, invalid domains, or inconsistent schemas. Historical retains `Churn`; current retains the same predictor columns and identifier. The clean files are exported to `outputs/`, leaving both raw CSV files unchanged.

```python
assert clean_historical.shape == (1200, 20)
assert clean_current.shape == (300, 19)
assert clean_historical.isna().sum().sum() == 0
assert clean_current.isna().sum().sum() == 0
assert clean_historical.duplicated().sum() == 0
assert clean_current.duplicated().sum() == 0
assert clean_historical["Customer_ID"].is_unique
assert clean_current["Customer_ID"].is_unique
assert clean_historical["Age"].between(18, 100).all() and clean_current["Age"].between(18, 100).all()
assert clean_historical["Satisfaction_Score"].between(1, 5).all() and clean_current["Satisfaction_Score"].between(1, 5).all()
assert (clean_historical[["Monthly_Income_JD", "Monthly_Spend_JD"]] >= 0).all().all()
assert (clean_current[["Monthly_Income_JD", "Monthly_Spend_JD"]] >= 0).all().all()
assert clean_historical["Churn"].isin([0, 1]).all()
assert clean_historical.drop(columns="Churn").columns.tolist() == clean_current.columns.tolist()

post_clean_summary = pd.DataFrame({
    "Historical": [len(clean_historical), clean_historical.shape[1], int(clean_historical.isna().sum().sum()), int(clean_historical.duplicated().sum()), int(clean_historical["Customer_ID"].duplicated().sum())],
    "Current": [len(clean_current), clean_current.shape[1], int(clean_current.isna().sum().sum()), int(clean_current.duplicated().sum()), int(clean_current["Customer_ID"].duplicated().sum())],
}, index=["Rows", "Columns", "Missing values", "Exact duplicate rows", "Duplicate-ID occurrences"])
display(post_clean_summary)

output_dir = Path("outputs")
output_dir.mkdir(exist_ok=True)
historical_output = output_dir / "Clean_Historical_Customers.csv"
current_output = output_dir / "Clean_Current_Customers.csv"
clean_historical.to_csv(historical_output, index=False)
clean_current.to_csv(current_output, index=False)

print("Exported:", historical_output)
print("Exported:", current_output)
print("All Task 2 post-cleaning checks passed.")
```
